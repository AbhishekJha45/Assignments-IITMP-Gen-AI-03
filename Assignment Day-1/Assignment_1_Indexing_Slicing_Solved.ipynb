# Day 1 — NumPy Indexing & Slicing (3D and 2D arrays)
Every answer is a single indexing/slicing statement; the expected values from the question sheet are checked with `assert`.

```python
import numpy as np
arr = np.arange(10, 70).reshape(3, 4, 5)
# the question sheet's array: 3 matrices, 4 rows, 5 columns, values 10..69 in order
print(arr.shape); print(arr[0])
```

## 3D array — Section A: individual elements

```python
print(arr[0,0,0], arr[0,2,4], arr[1,1,2], arr[1,2,3], arr[2,0,4], arr[2,3,1], arr[2,1,3], arr[1,3,4])
assert [arr[0,0,0], arr[0,2,4], arr[1,1,2], arr[1,2,3], arr[2,0,4], arr[2,3,1], arr[2,1,3], arr[1,3,4]] == [10,24,37,43,54,66,58,49]
```

## Section B: complete rows

```python
print(arr[0,1,:]); print(arr[1,2,:]); print(arr[2,3,:])
assert arr[0,1,:].tolist()==[15,16,17,18,19] and arr[1,2,:].tolist()==[40,41,42,43,44] and arr[2,3,:].tolist()==[65,66,67,68,69]
```

## Section C: complete 2D matrices

```python
print(arr[0]); print(arr[1]); print(arr[2])
```

## Section D: column extraction

```python
print(arr[0,:,0]); print(arr[1,:,2]); print(arr[2,:,4])
assert arr[0,:,0].tolist()==[10,15,20,25] and arr[1,:,2].tolist()==[32,37,42,47] and arr[2,:,4].tolist()==[54,59,64,69]
```

## Section E: slicing

```python
print('first two matrices\n', arr[:2].shape)
print('last two matrices\n', arr[1:].shape)
print('first two rows of every matrix\n', arr[:, :2, :])
print('last two rows of every matrix\n', arr[:, 2:, :])
print('first three columns of every matrix\n', arr[:, :, :3])
print('last two columns of every matrix\n', arr[:, :, 3:])
```

## Section F: advanced slicing

```python
print(arr[0, :2, :3]); print(arr[1, 2:, 2:]); print(arr[2, 1:, :2]); print(arr[0, :2, 3:])
assert arr[0,:2,:3].tolist()==[[10,11,12],[15,16,17]]
assert arr[1,2:,2:].tolist()==[[42,43,44],[47,48,49]]
assert arr[2,1:,:2].tolist()==[[55,56],[60,61],[65,66]]
assert arr[0,:2,3:].tolist()==[[13,14],[18,19]]
```

## Section G: single-statement questions

```python
print('alternate matrices', arr[::2].shape)
print('alternate rows', arr[:, ::2, :].shape)
print('alternate columns', arr[:, :, ::2].shape)
print('first column of every matrix\n', arr[:, :, 0])
print('last column of every matrix\n', arr[:, :, -1])
print('middle row of every matrix (index 2)\n', arr[:, 2, :])
print('middle column of every matrix (index 2)\n', arr[:, :, 2])
assert arr[::-1][0].tolist()==arr[2].tolist()          # reversed matrices
assert arr[:, ::-1, :][0,0].tolist()==arr[0,3].tolist() # reversed rows
assert arr[:, :, ::-1][0,0].tolist()==[14,13,12,11,10]  # reversed columns
```

## 2D array

```python
a = np.array([[10,11,12,13,14,15],[20,21,22,23,24,25],[30,31,32,33,34,35],
              [40,41,42,43,44,45],[50,51,52,53,54,55],[60,61,62,63,64,65]])
print(a.shape)
print(1, a[::2, :].tolist())
print(2, a[:, ::2].tolist())
print(3, a[:, 0].tolist()); print(4, a[:, -1].tolist())
print(5, a[3, :].tolist()); print(6, a[:, 3].tolist())
print(7, a[:3, :].tolist()); print(8, a[3:, :].tolist())
print(9, a[:, :3].tolist()); print(10, a[:, 3:].tolist())
print(11, a[::-1, :].tolist()); print(12, a[:, ::-1].tolist()); print(13, a[::-1, ::-1].tolist())
assert a[3,:].tolist()==[40,41,42,43,44,45] and a[:,3].tolist()==[13,23,33,43,53,63]
assert a[::-1,::-1][0].tolist()==[65,64,63,62,61,60]
```

### Write a single statement to extract

```python
print(a[1:4, 2:5].tolist())     # [[22,23,24],[32,33,34],[42,43,44]]
print(a[::2, 1::2].tolist())    # [[11,13,15],[31,33,35],[51,53,55]]
print(a[3:5, :].tolist())       # [[40..45],[50..55]]
print(a[:, 4:].tolist())        # last two columns, all rows
assert a[1:4,2:5].tolist()==[[22,23,24],[32,33,34],[42,43,44]]
assert a[::2,1::2].tolist()==[[11,13,15],[31,33,35],[51,53,55]]
assert a[3:5,:].tolist()==[[40,41,42,43,44,45],[50,51,52,53,54,55]]
assert a[:,4:].tolist()==[[14,15],[24,25],[34,35],[44,45],[54,55],[64,65]]
```
