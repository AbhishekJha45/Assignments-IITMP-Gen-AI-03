```python
employees = {
    101: "John",
    102: "Alice",
    103: "Peter",
    104: "Maria"
}

# Q1: Employee ID 102 exists
print(102 in employees)

# Q2: Employee ID 110 exists
print(110 in employees)

# Q3: Suitable message if the employee ID is not found
id_to_check = 110
if id_to_check in employees:
    print(f"Employee {id_to_check} found")
else:
    print(f"Employee {id_to_check} not found")
```

```python
grades = {
    "John": "A",
    "Alice": "B",
    "Peter": "A",
    "David": "C"
}

# Q6a: Students with grade A
print([name for name, grade in grades.items() if grade == "A"])

# Q6b: Students with grade C
print([name for name, grade in grades.items() if grade == "C"])
```

```python
runs = {
    "Virat": 105,
    "Rohit": 78,
    "Gill": 92,
    "Rahul": 78
}

# Q8a: Players who scored 78
print([name for name, score in runs.items() if score == 78])

# Q8b: Player who scored 105
print([name for name, score in runs.items() if score == 105])
```

```python
customers = {
    "John": "9876543210",
    "Alice": None,
    "Bob": "9988776655",
    "David": None
}

# Q9a: Alice phone available
print(customers["Alice"] is not None)

# Q9b: Customers with missing phone
print([name for name, phone in customers.items() if phone is None])

# Q9c: Count missing phones
print(sum(1 for phone in customers.values() if phone is None))
```

```python
emails = {
    "John": "john@gmail.com",
    "Peter": None,
    "Alice": "alice@gmail.com",
    "Mary": None
}

# Q10a: Employees with missing email
print([name for name, email in emails.items() if email is None])

# Q10b: Replace missing email
for name, email in emails.items():
    if email is None:
        emails[name] = "Not Available"
print(emails)
```

```python
books = {
    "Python": 15,
    "Java": 0,
    "C++": 8,
    "AI": 0
}

# Q11a: Books out of stock
print([title for title, qty in books.items() if qty == 0])

# Q11b: Count zero copies
print(sum(1 for qty in books.values() if qty == 0))
```

```python
patients = {
    "John": "",
    "Alice": "A+",
    "Peter": "",
    "Mary": "O-"
}

# Q12a: Patients missing blood group
print([name for name, group in patients.items() if not group])

# Q12b: Incomplete records count
print(sum(1 for group in patients.values() if not group))
```

```python
vehicles = {
    "TN01AB1234": "Car",
    "TN09XY5678": "Bike",
    "KA05CD1111": None,
    "KL07PQ2222": "Truck"
}

# Q13a: TN09XY5678 exists
print("TN09XY5678" in vehicles)

# Q13b: KA10AA0001 exists
print("KA10AA0001" in vehicles)

# Q13c: Missing information
print([reg for reg, vehicle in vehicles.items() if vehicle is None])

# Q13d: Count missing types
print(sum(1 for vehicle in vehicles.values() if vehicle is None))
```

```python
attendance = {
    "John": True,
    "Alice": False,
    "Bob": True,
    "David": None
}

# Q14a: Bob attendance available
print(attendance.get("Bob") is not None)

# Q14b: Attendance not recorded
print([name for name, value in attendance.items() if value is None])

# Q14c: Present students count
print(sum(1 for value in attendance.values() if value is True))

# Q14d: Absent students count
print(sum(1 for value in attendance.values() if value is False))
```

```python
ratings = {
    "Movie A": 4.5,
    "Movie B": None,
    "Movie C": 3.8,
    "Movie D": None
}

# Q15a: Movie C has rating
print(ratings.get("Movie C") is not None)

# Q15b: Movies with no ratings
print([movie for movie, rating in ratings.items() if rating is None])

# Q15c: Unrated movies count
print(sum(1 for rating in ratings.values() if rating is None))

# Q15d: Replace missing ratings with 0
for movie, rating in ratings.items():
    if rating is None:
        ratings[movie] = 0
print(ratings)
```
