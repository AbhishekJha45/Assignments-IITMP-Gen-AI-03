```python
from pathlib import Path
import pandas as pd

base_dir = Path.cwd()
required_files = [
    'patient_clinical_data_raw.csv',
    'ecommerce_orders_raw.csv',
    'iot_sensor_data_raw.csv',
    'customers.csv',
    'products.csv',
    'orders.csv'
]
if not all((base_dir / name).exists() for name in required_files):
    base_dir = base_dir.parent

base_dir
```

## Assignment 4: Customer and Order Analysis

Merge customer, product, and order data to compute revenue and customer insights.

```python
customers = pd.read_csv(base_dir / 'customers.csv')
products = pd.read_csv(base_dir / 'products.csv')
orders = pd.read_csv(base_dir / 'orders.csv')

customers['Customer_ID'] = customers['Customer_ID'].astype(str)
products['Product_ID'] = products['Product_ID'].astype(str)
orders['Customer_ID'] = orders['Customer_ID'].astype(str)
orders['Product_ID'] = orders['Product_ID'].astype(str)
orders['Order_Date'] = pd.to_datetime(orders['Order_Date'])

merged = orders.merge(customers, on='Customer_ID', how='left')
final = merged.merge(products, left_on='Product_ID', right_on='Product_ID', how='left')
final['Order_Value'] = final['Price'] * final['Quantity']

print('Total revenue:', round(final['Order_Value'].sum(), 2))
print('Average order value:', round(final['Order_Value'].mean(), 2))
print('\nRevenue by product category:')
print(final.groupby('Category')['Order_Value'].sum().sort_values(ascending=False))
print('\nTop 10 customers by spending:')
print(final.groupby('Customer_ID')['Order_Value'].sum().sort_values(ascending=False).head(10))
print('\nCity with highest revenue:')
print(final.groupby('City')['Order_Value'].sum().sort_values(ascending=False).head())
print('\nMonthly revenue:')
print(final.set_index('Order_Date').resample('ME')['Order_Value'].sum())
```
